A function that creates an engine

```python
def engine_creation(username,password,host,port,database_name):
    server_name = input('What is your server (postgresql : p)/(mysql : m) (p/m): '  )
    if server_name.lower() == 'p' :
        engine = create_engine(f"postgresql+psycopg2://{username}:{quote_plus(password)}@{host}:{port}/{database_name}")
    elif server_name.lower() == 'm' :
        engine = create_engine(f"mysql+pymysql://{username}:{quote_plus(password)}@{host}:{port}/{database_name}")
    else :
        print('Invalid response please enter p for postgreSQL and m for mySQL')
        return None
    return engine
```

Data Ingestor

```python
def ingestor(folder_path,engine_name):
    start = time.time()
    for file in os.listdir(folder_path):
        if file.endswith('.csv'):
            file_path = os.path.join(folder_path,file)
            table_name = file[:-4]
            print(f"Appending {file} into the table: {table_name}")
            df = pd.read_csv(file_path)
            add_file_to_db(df,table_name,engine_name)
            end = time.time()
            total_time = (end-start)/60
            print(f'Completed appending {file} into table: {table_name} shape = {df.shape}')
            print(f'Total time = {total_time} minutes')
```

Data transmitter

```python
def add_file_to_db(df,table_name,engine):
    df.to_sql(name=table_name,con=engine,if_exists='replace',index=False,chunksize=10000,method='multi')
```
